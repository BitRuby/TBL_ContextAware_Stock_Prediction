```python
import yfinance as yf
import numpy as np
import math
from tabulate import tabulate
from statsmodels.tsa.stattools import adfuller
import matplotlib.pyplot as plt
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from arch import arch_model
import pandas as pd
import plotly.express as px
import xgboost as xgb
from xgboost import plot_importance, plot_tree
from sklearn.metrics import mean_absolute_percentage_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import pandas_ta as ta
```

```python
dataset = yf.download('SPX', "2005-01-01", "2025-01-01")
```

```python
dataset.head()
```

```python
log_return = np.log(dataset['Close'] / dataset['Close'].shift(1))
dataset["sma50"] = dataset["Close"].rolling(window=50).mean()
dataset["sma10"] = dataset["Close"].rolling(window=10).mean()
dataset["rsi"] = ta.rsi(dataset["Close"]["SPX"], length=50)
dataset["rolling_std"] = abs(log_return.dropna().rolling(window=21, min_periods=21).std().dropna())
dataset["prev1"] = (dataset["rolling_std"] > dataset["rolling_std"].shift(1)).astype(int)
dataset["prev2"] = (dataset["rolling_std"] > dataset["rolling_std"].shift(2)).astype(int)
dataset["prev3"] = (dataset["rolling_std"] > dataset["rolling_std"].shift(3)).astype(int)
dataset["prev4"] = (dataset["rolling_std"] > dataset["rolling_std"].shift(4)).astype(int)
dataset["prev5"] = (dataset["rolling_std"] > dataset["rolling_std"].shift(5)).astype(int)
dataset["prev10"] = (dataset["rolling_std"] > dataset["rolling_std"].shift(10)).astype(int)
dataset["prev21"] = (dataset["rolling_std"] > dataset["rolling_std"].shift(21)).astype(int)
dataset["prev50"] = (dataset["rolling_std"] > dataset["rolling_std"].shift(50)).astype(int)
dataset["BB_Mid"] = dataset["Close"].rolling(window=20).mean()
dataset["BB_Std"] = dataset["Close"].rolling(window=20).std()
dataset["BB_Upper"] = dataset["BB_Mid"] + (dataset["BB_Std"] * 2)
dataset["BB_Lower"] = dataset["BB_Mid"] - (dataset["BB_Std"] * 2)
dataset["crossover"] = (dataset["sma10"] > dataset["sma50"]).astype(int)
dataset["Date"] = pd.to_datetime(dataset.index)
dataset['dayofweek'] = dataset["Date"].dt.dayofweek
dataset['quarter'] = dataset["Date"].dt.quarter
dataset['month'] = dataset["Date"].dt.month
dataset['year'] = dataset["Date"].dt.year
dataset['dayofyear'] = dataset["Date"].dt.dayofyear
dataset['dayofmonth'] = dataset["Date"].dt.day
dataset["rolling_std_binary"] = (dataset["rolling_std"] > dataset["rolling_std"].shift(-1)).astype(int)

dataset = dataset.dropna()
```

```python
plt.scatter(dataset["rolling_std_binary"],dataset["prev1"])
plt.xlabel('Train Close Lag-0')
plt.ylabel('Train Close Lag-1')
```

```python
X_train = dataset[["prev1"]][:-252]
y_train = dataset["rolling_std_binary"][:-252]
X_test = dataset[["prev1"]][-252:]
y_test = dataset["rolling_std_binary"][-252:]
```

```python
from sklearn.model_selection import TimeSeriesSplit
tscv = TimeSeriesSplit(n_splits=5)
for train_idx, test_idx in tscv.split(X_train):
    X_fold_train, X_fold_test = X_train.iloc[train_idx], X_train.iloc[test_idx]
    y_fold_train, y_fold_test = y_train.iloc[train_idx], y_train.iloc[test_idx]
```

```python
# reg = xgb.XGBRegressor(n_estimators=1000, early_stopping_rounds=50, objective="binary:logistic",  # Binary classification
#     eval_metric="logloss",  # Log loss for binary problems
#     use_label_encoder=False,
#     learning_rate=0.05,  # Step size shrinkage
#     max_depth=5,  # Maximum depth of trees
#     subsample=0.8,  # Subsample ratio for training data
#     colsample_bytree=0.8)  # Subsample ratio for features)
# reg.fit(X_train, y_train, eval_set=[(X_train, y_train), (X_test, y_test)], verbose=False)
# _ = plot_importance(reg, height=0.9)
model = xgb.XGBClassifier(
    objective="binary:logistic",  # Binary classification
    eval_metric="logloss",  
    use_label_encoder=False,
    n_estimators=200,
    learning_rate=0.005,
    max_depth=5,
    subsample=0.65,  # Reduce overfitting
    colsample_bytree=0.7,
    reg_lambda=10,  # L2 regularization
    reg_alpha=5  # L1 regularization
)
model.fit(X_fold_train, y_fold_train)
```

```python
predictions = model.predict(X_fold_test)
```

```python
# Compute accuracy
accuracy = accuracy_score(y_fold_test, predictions)
print(f"Accuracy: {accuracy:.4f}")

# Classification report (Precision, Recall, F1-score)
print("\nClassification Report:")
print(classification_report(y_fold_test, predictions))

# Confusion matrix
conf_matrix = confusion_matrix(y_fold_test, predictions)
print("\nConfusion Matrix:")
print(conf_matrix)
```
