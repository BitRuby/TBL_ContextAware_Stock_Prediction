```python
def keep_leading_letters(text):
    match = re.match(r'^[A-Z]+', text)
    return match.group(0) if match else ''

def sharpe_ratio(returns, risk_free_rate=0.0):
    if len(returns) < 2:
        return 0  # Avoid division by zero
    excess_returns = np.array(returns) - risk_free_rate
    return np.mean(excess_returns) / np.std(excess_returns) if np.std(excess_returns) > 0 else 0

def sortino_ratio(returns, risk_free_rate=0.0):
    excess_returns = np.array(returns) - risk_free_rate
    downside_returns = excess_returns[excess_returns < 0]
    downside_std = np.std(downside_returns) if len(downside_returns) > 0 else 0
    return np.mean(excess_returns) / downside_std if downside_std > 0 else 0

def remove_unwanted_hashtags(text, keep):
    def replacer(match):
        hashtag = match.group(0)[1:]
        if hashtag.lower() in keep:
            return match.group(0)
        else:
            return ''
    cleaned =  re.sub(r'[@$#]\w+', replacer, text)
    return re.sub(r'\s{2,}', ' ', cleaned).strip()

def apply_transformations(transformations, text):
    if "REMOVE_USERNAMES" in transformations:
        remove_usernames_pattern = r'@\S+'
        text = re.sub(remove_usernames_pattern, "", text)
    if "REMOVE_URLS" in transformations:
        remove_urls_pattern = r'http\S+'
        text = re.sub(remove_urls_pattern, "", text)
    if "REMOVE_PUNCTUATION_MARKS" in transformations:
        remove_pun_pattern_1 = r"(?<!\d)\.(?!\d)|[^\w\s.']"
        remove_pun_pattern_2 = r"'"
        remove_pun_pattern_3 = r"\s\s+"
        sub1 = re.sub(remove_pun_pattern_1, " ", text)
        sub2 = re.sub(remove_pun_pattern_2, "", sub1)
        text = re.sub(remove_pun_pattern_3, " ", sub2)
    if "TEXT_TO_LOWER" in transformations:
        text = text.lower()
    if "REMOVE_SHORT_WORDS" in transformations:
        remove_short_pattern = r'\b\w{1,2}\b'
        text = re.sub(remove_short_pattern, '', text)
    if "REMOVE_ENDING_HASHTAGS" in transformations:
        pattern = r'([#$@\$]\w+)(?=(\s[#$@\$]\w+)*\s*$)'
        text = re.sub(pattern, '', text)
    if "REMOVE_HEX" in transformations:
        text = re.sub(r'\b0x[a-fA-F0-9]{6,}\b', '', text)
    if "REMOVE_EMOTICONS" in transformations:
        emoji_pattern = re.compile(
            "[" 
            u"\U0001F600-\U0001F64F"  # Emoticons
            u"\U0001F300-\U0001F5FF"  # Symbols & pictographs
            u"\U0001F680-\U0001F6FF"  # Transport & map
            u"\U0001F1E0-\U0001F1FF"  # Flags
            u"\U00002700-\U000027BF"  # Dingbats
            u"\U0001F900-\U0001F9FF"  # Supplemental Symbols & Pictographs (includes 🤝)
            u"\U00002600-\U000026FF"  # Misc symbols (e.g. ☀️☂️)
            u"\U00002B00-\U00002BFF"  # Arrows etc.
            u"\U0001FA70-\U0001FAFF"  # Symbols and Pictographs Extended-A
            "]+", flags=re.UNICODE
        )
        text = emoji_pattern.sub(r'', text)
    return re.sub(r"\s{2,}", " ", text).strip()
def rsi_to_label(rsi_series):
    return np.where(rsi_series < 30, LABEL_BULLISH, np.where(rsi_series > 70, LABEL_BEARISH, LABEL_NEUTRAL))

def calculate_barriers(price, volatility, Fu, Fl):
    upper = price + price * volatility * Fu
    lower = price - price * volatility * Fl
    return upper, lower

def get_best_params_for_day(results, day):
    day = pd.to_datetime(day)
    for res in results:
        if res['start'] <= day < res['end']:
            return res['best_params']
    return None  # If day is outside all defined windows

def get_label_for_a_day(labels, day):
    day = pd.to_datetime(day)
    for label in labels:
        if label['start'] <= day < label['end']:
            return label['label']
    return None  # If day is outside all defined windows

def get_prev_label_for_a_day(labels, day):
    day = pd.to_datetime(day)
    for label in labels:
        if label['start'] <= day < label['end']:
            return label['prev_label']
    return None  # Ensure it's always a tuple with 2 values

def assign_labels(data, fu, fl, vt, best_params_for_day=None):
    barriers = None
    start_day = None
    end_day = None
    prev_label = None
    labels = []
    for index, day in data[:-1].iterrows():
        if best_params_for_day != None:
            params = get_best_params_for_day(best_params_for_day, day["date"])
            if params is not None:
                fu, fl, vt = params
            else:
                fu, fl, vt = 1.0, 1.0, 10
        if barriers == None:
            barriers = calculate_barriers(day["Open"], day["Volatility"], fu, fl)
            high_barrier, low_barrier = barriers
            start_day = day["date"]
            days = 0
        days += 1
        if day["High"] > barriers[0]:
            labels.append({"start": start_day, "end": data["date"][index+1], "label": LABEL_BULLISH, "prev_label": prev_label})
            start_day = None
            end_day = None
            barriers = None
            prev_label = LABEL_BULLISH
        elif day["Low"] < barriers[1]:
            labels.append({"start": start_day, "end": data["date"][index+1], "label": LABEL_BEARISH, "prev_label": prev_label})
            start_day = None
            end_day = None
            barriers = None
            prev_label = LABEL_BEARISH
        elif start_day != None and end_day == None and days == vt:
            labels.append({"start": start_day, "end": data["date"][index+1], "label": LABEL_NEUTRAL, "prev_label": prev_label})
            start_day = None
            end_day = None
            barriers = None
            prev_label = LABEL_NEUTRAL
            days = 0
    return labels

def simulate_returns(df, labels, Vt):
    returns = []
    for i in range(len(df) - Vt):
        start_price = df['Close'].iloc[i]
        end_price = df['Close'].iloc[i+Vt]
        label = get_label_for_a_day(labels, df['date'].iloc[i])
        if label == LABEL_BULLISH:
            ret = (end_price - start_price) / start_price
        elif label == LABEL_BEARISH:
            ret = (start_price - end_price) / start_price
        else:
            ret = 0
        returns.append(ret)
    returns.extend([0] * Vt)
    return np.array(returns)

def optimize_sharpe(df, Fu_range, Fl_range, Vt_range, rf=0.04):
    df = df.sort_values("date").reset_index(drop=True)
    df["date"] = pd.to_datetime(df["date"])
    
    start_date = df["date"].min()
    end_date = df["date"].max()
    window_results = []

    current_start = start_date

    while current_start + relativedelta(months=6) <= end_date:
        current_end = current_start + relativedelta(months=6)
        window_df = df[(df["date"] >= current_start) & (df["date"] < current_end)].reset_index(drop=True)

        best_sr = -np.inf
        best_params = None

        for Fu in Fu_range:
            for Fl in Fl_range:
                for Vt in Vt_range:
                    if len(window_df) < Vt + 1:
                        continue
                    labels = assign_labels(window_df, Fu, Fl, Vt)
                    rets = simulate_returns(window_df, labels, Vt)
                    mean_ret = np.mean(rets) * 252
                    std_ret = np.std(rets) * np.sqrt(252)
                    if std_ret > 0:
                        sr = (mean_ret - rf) / std_ret
                        if sr > best_sr:
                            best_sr = sr
                            best_params = (Fu, Fl, Vt)
        
        window_results.append({
            "start": current_start,
            "end": current_end,
            "best_params": best_params,
            "sharpe": best_sr
        })

        current_start += relativedelta(months=6)

    return window_results

# Run optimization
Fu_range = np.arange(0.5, 3.1, 0.5)
Fl_range = np.arange(0.5, 3.1, 0.5)
Vt_range = range(8, 16)
```

```python
cases = pd.read_csv("TBL_Test_Cases.csv")
# for test_id in range(len(cases)):
test_id = 2
NAME = cases["Name"][test_id]
MARKET_NAME = cases["Market"][test_id]
TWEETS_ID = cases["TweetsId"][test_id]
MODEL = cases["Model"][test_id]
ENGAGEMENT_POSTS = cases["EngagementPosts"][test_id]
data = yf.download(MARKET_NAME, start=STARTING_DATE, end=ENDING_DATE).reset_index()
data.columns = data.columns.get_level_values(0)
data["Date"] = pd.to_datetime(data["Date"])
data = data[["Date", "Close", "Open", "High", "Low"]]
data['LogReturn'] = np.log(data['Close'] / data['Close'].shift(1))
data['Volatility'] = data['LogReturn'].ewm(span=30).std()
data["RSI"] = ta.rsi(data["Close"], length=14)
data["RSI_label"] = rsi_to_label(data["RSI"])
data["ROC"] = data['Close'].pct_change(periods=8) * 100
roc_std = data['ROC'].rolling(window=8).std()
data['ROC_label'] = np.where(data['ROC'] > roc_std, LABEL_NEUTRAL, np.where(data['ROC'] < -roc_std, LABEL_BEARISH, LABEL_NEUTRAL))
data = data.dropna()
data["date"] = pd.to_datetime(data["Date"]).dt.date

optimized = optimize_sharpe(data, Fu_range, Fl_range, Vt_range)
labels = assign_labels(data, 1,1,1, optimized)
data["label"] = data.apply(lambda x: get_label_for_a_day(labels, x["date"]), axis=1)
data["previous_label"]  = data.apply(lambda x: get_prev_label_for_a_day(labels, x["date"]), axis=1)

data = data.dropna()

# tweets_aggregated = pd.DataFrame(tweets.find({
#     "tag_id": ObjectId(TWEETS_ID),
#     "created_at": {
#         "$gte": datetime.strptime(STARTING_DATE, "%Y-%m-%d"),
#         "$lte": datetime.strptime(ENDING_DATE, "%Y-%m-%d").replace(hour=23, minute=59, second=59, microsecond=999999)
#     },
# #     "likes": { "$gt": 20 }
# }))
tweets_aggregated = tweets_aggregated[tweets_aggregated["tag_id"] == TWEETS_ID]

if ENGAGEMENT_POSTS == True:
    tweets_aggregated = tweets_aggregated[tweets_aggregated["likes"] > 20]

tweets_aggregated["content"] = tweets_aggregated["content"].apply(lambda x: apply_transformations(TRANSFORMATIONS, remove_unwanted_hashtags(x, keep_leading_letters(MARKET_NAME))))
tweets_aggregated['date'] = pd.to_datetime(tweets_aggregated['created_at'])
tweets_aggregated = tweets_aggregated.sort_values(by='date')
tweets_aggregated['date'] = tweets_aggregated['date'].dt.date
if RANDOM_SAMPLE == True:  
    tweets_aggregated = tweets_aggregated.groupby('date', group_keys=False).apply(
        lambda x: x.sample(n=min(len(x), DAILY_SAMPLE_SIZE), random_state=42)
    )
merged = pd.merge(tweets_aggregated, data, on='date', suffixes=('_df1', '_df2'))
merged = merged.dropna()

split_index = int(len(merged) * 0.8)
train_data =  merged[:split_index]
test_data = merged[split_index:]
#     # tokenizer = BertTokenizer.from_pretrained(MODEL_NAME) For FinBERT
#     tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME) #For CryptoBERT

tokenizer = AutoTokenizer.from_pretrained(MODEL) if MODEL == "ElKulako/cryptobert" else BertTokenizer.from_pretrained(MODEL)

label_encoder = LabelEncoder()
targets = label_encoder.fit_transform(train_data['label'])  # Maps up/down/neutral to 0/1/2
# Create prompt-based inputs
def create_prompt(tweet, prev_label, roc, rsi):
    return f"Previous Label: {prev_label}, ROC: {roc}, RSI: {rsi}, Tweet: {tweet}"

prompts = [create_prompt(t, pl, r, rs) for t, pl, r, rs in zip(train_data['content'], train_data['previous_label'], train_data['ROC_label'], train_data['RSI_label'])]

kf = KFold(n_splits=FOLDS, shuffle=True, random_state=42)

best_val_acc = 0
best_fold = -1
best_model = None  # Optional: to keep best model in memory
tr_idx = []
va_idx = []

for fold, (train_idx, val_idx) in enumerate(kf.split(prompts)):
    print(f"\n=== Fold {fold + 1} ===")
    train_prompts = [prompts[i] for i in train_idx]
    val_prompts = [prompts[i] for i in val_idx]
    train_labels = tf.convert_to_tensor(targets[train_idx])
    val_labels = tf.convert_to_tensor(targets[val_idx])
    # Tokenize
    train_enc = tokenizer(train_prompts, truncation=True, padding=True, max_length=128, return_tensors="tf")
    val_enc = tokenizer(val_prompts, truncation=True, padding=True, max_length=128, return_tensors="tf")
    # Prepare dataset
    train_dataset = tf.data.Dataset.from_tensor_slices((
        {
            "input_ids": train_enc["input_ids"],
            "attention_mask": train_enc["attention_mask"],
            "token_type_ids": train_enc.get("token_type_ids", tf.zeros_like(train_enc["input_ids"]))  # for compatibility
        },
        train_labels
    )).shuffle(100).batch(BATCH_SIZE)

    val_dataset = tf.data.Dataset.from_tensor_slices((
        {
            "input_ids": val_enc["input_ids"],
            "attention_mask": val_enc["attention_mask"],
            "token_type_ids": val_enc.get("token_type_ids", tf.zeros_like(val_enc["input_ids"]))  # for compatibility
        },
        val_labels
    )).batch(BATCH_SIZE)

#     #     model = TFBertForSequenceClassification.from_pretrained(MODEL_NAME, num_labels=NUM_LABELS) For FinBERT
#         model = TFAutoModelForSequenceClassification.from_pretrained(MODEL_NAME, num_labels = NUM_LABELS, from_pt=True) #for CryptoBERT
    model = TFAutoModelForSequenceClassification.from_pretrained(MODEL, num_labels = NUM_LABELS, from_pt=True) if MODEL == "ElKulako/cryptobert" else TFBertForSequenceClassification.from_pretrained(MODEL, num_labels=NUM_LABELS)  

    if MODEL == "ElKulako/cryptobert":
        for layer in model.roberta.encoder.layer[:11]:
            layer.trainable = False
    else:
        for layer in model.bert.encoder.layer[:11]:
            layer.trainable = False
    # # Optimizer & learning rate scheduler
    num_train_steps = len(train_data) * EPOCHS
    num_warmup_steps = int(0.1 * num_train_steps)

    optimizer, lr_schedule = create_optimizer(
        init_lr=LEARNING_RATE,
        num_train_steps=num_train_steps,
        num_warmup_steps=num_warmup_steps
    )

    # Compile model
    model.compile(
        optimizer=optimizer,
        loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
        metrics=["accuracy"]
    )

    # Train
    history = model.fit(train_dataset,validation_data=val_dataset, epochs=EPOCHS)

    val_acc = history.history["val_accuracy"][-1]
    print(f"Fold {fold + 1} validation accuracy: {val_acc:.4f}")

    if val_acc > best_val_acc:
        best_val_acc = val_acc
        best_fold = fold
        best_model = model
        tr_idx = train_idx
        va_idx = val_idx
    else:
        # Free memory of model not being kept
        del model
        tf.keras.backend.clear_session()
        import gc
        gc.collect()

print(f"\n✅ Best Fold: {best_fold + 1} with validation accuracy: {best_val_acc:.4f}")
print("Training complete.")

test_prompts = [
    create_prompt(t, pl, r, rs)
    for t, pl, r, rs in zip(
        test_data['content'],
        test_data['previous_label'],
        test_data['ROC_label'],
        test_data['RSI_label']
    )
]

test_enc = tokenizer(
    test_prompts,
    truncation=True,
    padding=True,
    max_length=128,
    return_tensors="tf"
)

test_dataset = tf.data.Dataset.from_tensor_slices({
    "input_ids": test_enc["input_ids"],
    "attention_mask": test_enc["attention_mask"],
    "token_type_ids": test_enc.get("token_type_ids", tf.zeros_like(test_enc["input_ids"]))
}).batch(BATCH_SIZE)

# Predict logits
pred_logits = best_model.predict(test_dataset).logits

# Convert to predicted class labels
pred_classes = tf.argmax(pred_logits, axis=1).numpy()

# Decode labels back to original strings (e.g., 'up', 'down', etc.)
pred_labels = label_encoder.inverse_transform(pred_classes)

true_labels = label_encoder.transform(test_data["label"])

accuracy = accuracy_score(true_labels, pred_classes)

print("Test Accuracy:", accuracy)

print(classification_report(true_labels, pred_classes, labels=[0, 1, 2], target_names=label_encoder.classes_))
f1 = f1_score(true_labels, pred_classes, average="weighted")  # or "macro", "micro"
print(f"F1 Score: {f1:.4f}")

test_df = test_data.copy()
test_df = test_df.reset_index(drop=True)
test_df["pred_label"] = pred_classes
test_df["true_label_encoded"] = label_encoder.transform(test_df["label"])

# Convert predicted labels to signals
label_to_signal = {0: -1, 1: 0, 2: 1}
test_df["signal"] = test_df["pred_label"].map(label_to_signal)

# Group by date and get majority vote (mode) for signal
daily_signals = test_df.groupby("date")["signal"].agg(lambda x: mode(x)[0][0])

daily_close = test_df.groupby("date")["Close"].last()

daily_df = pd.DataFrame({
    "signal": daily_signals,
    "Close": daily_close
})

profits = []
purchases = 0
sells = 0
successes = 0
fails = 0

for i in range(1, len(daily_df)):
    prev_price = daily_df.iloc[i - 1]["Close"]
    curr_price = daily_df.iloc[i]["Close"]
    action = daily_df.iloc[i - 1]["signal"]

    price_change = curr_price - prev_price

    if action == 0:  # Buy
        reward = price_change
        purchases += 1
    elif action == 1:  # Sell
        reward = -price_change
        sells += 1
    else:  # Hold
        reward = 0

    successes = successes + 1 if reward > 0 else successes
    fails = fails + 1 if reward <= 0 else fails

    profits.append(reward)


sharpe = sharpe_ratio(profits)
sortino = sortino_ratio(profits)
total_trades = len(profits)
win_rate = (successes / total_trades) * 100 if total_trades > 0 else 0
total_profits = sum(profits)
profit_per_trade = total_profits / total_trades if total_trades > 0 else 0

# Start cumulative return with 0 for first day
cumulative_returns = [0] + list(np.cumsum(profits))
print(f"Cumulative Return: {cumulative_returns}")
columns = ["Test Name", "# Purchases", "# Sells", "Successes", "Fails", "Sortino Ratio", "Sharpe Ratio", "Profit Per Trade", "Win Rate", "Accuracy", "F1Score", "Cumulative Returns"]

pd.DataFrame([[NAME, purchases, sells, successes, fails, sortino, sharpe, profit_per_trade, win_rate, accuracy, f1, cumulative_returns]], columns=columns).to_csv(f"Test_Results.csv", mode="a", index=False, header=not os.path.exists("TBL_Test_Results.csv"))
```

```python
from collections import Counter
elements = []
for group in merged.groupby("Date"):
    all_prompts = [
    create_prompt(t, pl, r, rs)
    for t, pl, r, rs in zip(
        group[1]['content'],
        group[1]['previous_label'],
        group[1]['ROC_label'],
        group[1]['RSI_label']
        )
    ]
    all_enc = tokenizer(
        all_prompts,
        truncation=True,
        padding=True,
        max_length=128,
        return_tensors="tf"
    )
    
    all_dataset = tf.data.Dataset.from_tensor_slices({
        "input_ids": all_enc["input_ids"],
        "attention_mask": all_enc["attention_mask"],
        "token_type_ids": all_enc.get("token_type_ids", tf.zeros_like(all_enc["input_ids"]))
    }).batch(BATCH_SIZE)

    # Predict logits
    pred_logits = best_model.predict(all_dataset).logits
    pred_classes = tf.argmax(pred_logits, axis=1).numpy()

    pred_labels = label_encoder.inverse_transform(pred_classes)
    counter = Counter(pred_labels)
    most_common = counter.most_common(1)[0][0]
    elements.append({"date": group[0], "predicted_label": most_common})

pd.DataFrame(elements).to_csv("predicted_labels.csv")
```
